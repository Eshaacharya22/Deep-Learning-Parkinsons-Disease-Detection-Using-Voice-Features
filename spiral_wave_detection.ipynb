```python
import os
import zipfile

zip_path = '/content/archive (3).zip'
extract_to = '/content/unzipped'

def is_zipfile_valid(filepath):
    if not os.path.exists(filepath):
        print(f"Error: File does not exist: {filepath}")
        return False
    if os.path.getsize(filepath) == 0:
        print(f"Error: File is empty: {filepath}")
        return False
    with open(filepath, 'rb') as f:
        signature = f.read(4)
    if signature != b'PK\x03\x04':
        print("Error: File signature does not match a ZIP file.")
        return False
    if not zipfile.is_zipfile(filepath):
        print("Error: File is not recognized as a valid zip file by zipfile module.")
        return False
    return True

def unzip_file(zip_path, extract_to):
    print("Current working directory:", os.getcwd())
    print(f"Checking ZIP file at: {zip_path}")
    if not is_zipfile_valid(zip_path):
        return
    os.makedirs(extract_to, exist_ok=True)
    try:
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
        print(f"Unzipped files to: {extract_to}")
    except zipfile.BadZipFile:
        print("Error: The file is not a valid zip archive or is corrupted.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Run the unzip function
unzip_file(zip_path, extract_to)
```

```python
# import os
# import zipfile
# import tarfile
# from tensorflow.keras.preprocessing.image import ImageDataGenerator

# archive_path = '/content/archive (3).zip'
# extract_to = '/content/dataset'

# # -----------------------------
# # 1. Check file
# # -----------------------------
# if not os.path.exists(archive_path):
#     raise FileNotFoundError(f"❌ File not found: {archive_path}")

# print(f"✅ Found file: {archive_path}")
# print(f"📦 File size: {os.path.getsize(archive_path)} bytes")

# # -----------------------------
# # 2. Extract if needed
# # -----------------------------
# if not os.path.exists(extract_to):
#     os.makedirs(extract_to, exist_ok=True)
#     print(f"\n📂 Extracting {archive_path} to {extract_to}...")

#     if zipfile.is_zipfile(archive_path):
#         with zipfile.ZipFile(archive_path, 'r') as zip_ref:
#             zip_ref.extractall(extract_to)
#         print("✅ ZIP extraction complete.")
#     elif tarfile.is_tarfile(archive_path):
#         with tarfile.open(archive_path, 'r:*') as tar_ref:
#             tar_ref.extractall(extract_to)
#         print("✅ TAR extraction complete.")
#     else:
#         print("⚠️ Not zip/tar. Trying 7z extraction...")
#         os.system(f'7z x "{archive_path}" -o{extract_to} -y')
#         print("✅ 7z extraction complete.")
# else:
#     print(f"📂 Dataset already extracted in {extract_to}")

# # -----------------------------
# # 3. Detect structure
# # -----------------------------
# print("\n📂 Extracted folder contents:")
# print(os.listdir(extract_to))

# base_dir = os.path.join(extract_to, 'drawings')

# # case 1: if training/testing already exist
# train_dir = os.path.join(base_dir, 'training')
# test_dir = os.path.join(base_dir, 'testing')

# if not os.path.exists(train_dir) or not os.path.exists(test_dir):
#     print("⚠️ No training/testing folders found.")
#     print("📂 drawings contains:", os.listdir(base_dir))

#     # fallback: maybe drawings directly has class folders
#     train_dir = base_dir
#     test_dir = base_dir
#     print("👉 Using drawings as both train/test for now.")

# # -----------------------------
# # 4. ImageDataGenerator
# # -----------------------------
# datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

# spiral_train = datagen.flow_from_directory(
#     train_dir,
#     target_size=(256, 256),
#     batch_size=32,
#     class_mode='binary',
#     subset='training'
# )

# spiral_test = datagen.flow_from_directory(
#     test_dir,
#     target_size=(256, 256),
#     batch_size=32,
#     class_mode='binary',
#     subset='validation'
# )

# print("\n✅ Data generators ready!")



import os
import zipfile
import tarfile
from tensorflow.keras.preprocessing.image import ImageDataGenerator

archive_path = '/content/archive (3).zip'
extract_to = '/content/dataset'

# 1. Check file
if not os.path.exists(archive_path):
    raise FileNotFoundError(f"❌ File not found: {archive_path}")

print(f"✅ Found file: {archive_path}")
print(f"📦 File size: {os.path.getsize(archive_path)} bytes")

# 2. Extract if needed
if not os.path.exists(extract_to):
    os.makedirs(extract_to, exist_ok=True)
    print(f"\n📂 Extracting {archive_path} to {extract_to}...")

    if zipfile.is_zipfile(archive_path):
        with zipfile.ZipFile(archive_path, 'r') as zip_ref:
            zip_ref.extractall(extract_to)
        print("✅ ZIP extraction complete.")
    elif tarfile.is_tarfile(archive_path):
        with tarfile.open(archive_path, 'r:*') as tar_ref:
            tar_ref.extractall(extract_to)
        print("✅ TAR extraction complete.")
    else:
        print("⚠️ Not zip/tar. Trying 7z extraction...")
        os.system(f'7z x "{archive_path}" -o{extract_to} -y')
        print("✅ 7z extraction complete.")
else:
    print(f"📂 Dataset already extracted in {extract_to}")

print("\n📂 Extracted folder contents:")
print(os.listdir(extract_to))

# 3. Define dataset paths based on your structure
datasets = {
    'drawings_spiral': os.path.join(extract_to, 'drawings', 'spiral'),
    'drawings_wave': os.path.join(extract_to, 'drawings', 'wave'),
    'spiral': os.path.join(extract_to, 'spiral'),
    'wave': os.path.join(extract_to, 'wave')
}

# 4. Function to create generators if train/test folders exist
def create_generators(dataset_path, target_size=(256,256), batch_size=32, class_mode='binary'):
    train_dir = os.path.join(dataset_path, 'training')
    test_dir = os.path.join(dataset_path, 'testing')

    if not os.path.exists(train_dir) or not os.path.exists(test_dir):
        raise FileNotFoundError(f"Training/testing folders not found in {dataset_path}")

    datagen_train = ImageDataGenerator(rescale=1./255)
    datagen_test = ImageDataGenerator(rescale=1./255)

    train_gen = datagen_train.flow_from_directory(
        train_dir,
        target_size=target_size,
        batch_size=batch_size,
        class_mode=class_mode,
        shuffle=True
    )

    test_gen = datagen_test.flow_from_directory(
        test_dir,
        target_size=target_size,
        batch_size=batch_size,
        class_mode=class_mode,
        shuffle=False
    )

    return train_gen, test_gen

# 5. Create all generators
data_generators = {}
for name, path in datasets.items():
    try:
        train_gen, test_gen = create_generators(path)
        data_generators[name] = {'train': train_gen, 'test': test_gen}
        print(f"✅ Generators ready for dataset: {name}")
    except FileNotFoundError as e:
        print(f"⚠️ Skipping dataset {name}: {e}")

print("\n✅ All data generators created!")
```

```python
# =============================
# 1. Import Libraries
# =============================
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt

# =============================
# 2. Define Dataset Paths
# =============================
base_dir = "/content/dataset/drawings"   # ✅ corrected
train_dir = os.path.join(base_dir, "spiral", "training")
test_dir = os.path.join(base_dir, "spiral", "testing")

if not os.path.exists(train_dir) or not os.path.exists(test_dir):
    raise FileNotFoundError(f"❌ Training/Testing folders not found.\n{train_dir}\n{test_dir}")

print("✅ Dataset folders found!")
print("Train classes:", os.listdir(train_dir))
print("Test classes:", os.listdir(test_dir))

# =============================
# 3. Data Generators (with Augmentation)
# =============================
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode="nearest"
)

test_datagen = ImageDataGenerator(rescale=1./255)

spiral_train = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=16,
    class_mode='binary'
)

spiral_test = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=16,
    class_mode='binary'
)

# =============================
# 4. Build Transfer Learning Model
# =============================
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False  # freeze pretrained weights

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu', kernel_regularizer='l2'),
    layers.Dropout(0.6),
    layers.Dense(1, activation='sigmoid')  # binary classification
])

model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# =============================
# 5. Callbacks
# =============================
callbacks = [
    EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True),
    ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=4, verbose=1)
]

# =============================
# 6. Train the Model
# =============================
history = model.fit(
    spiral_train,
    validation_data=spiral_test,
    epochs=40,
    callbacks=callbacks
)

# =============================
# 7. Evaluate Model
# =============================
loss, acc = model.evaluate(spiral_test, verbose=0)
print(f"\n✅ Final Test Accuracy: {acc:.2f}")

# =============================
# 8. Plot Training Curves
# =============================
plt.figure(figsize=(12,5))

# Accuracy
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label="Train Acc")
plt.plot(history.history['val_accuracy'], label="Val Acc")
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

# Loss
plt.subplot(1,2,2)
plt.plot(history.history['loss'], label="Train Loss")
plt.plot(history.history['val_loss'], label="Val Loss")
plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()
```

```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt

def prepare_and_train(dataset_name):
    print(f"\n=== Training on dataset: {dataset_name} ===")
    base_dir = f"/content/dataset/drawings/{dataset_name}"
    train_dir = os.path.join(base_dir, "training")
    test_dir = os.path.join(base_dir, "testing")

    # Check directories exist
    if not os.path.exists(train_dir) or not os.path.exists(test_dir):
        raise FileNotFoundError(f"Training/testing folders missing for {dataset_name}")

    print("Train classes:", os.listdir(train_dir))
    print("Test classes:", os.listdir(test_dir))

    # Data generators with augmentation for train, and only rescale for test
    # train_datagen = ImageDataGenerator(
    #     rescale=1./255,
    #     rotation_range=20,
    #     width_shift_range=0.2,
    #     height_shift_range=0.2,
    #     shear_range=0.2,
    #     zoom_range=0.2,
    #     horizontal_flip=True,
    #     brightness_range=[0.8, 1.2],
    #     fill_mode="nearest"
    # )



    train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=10,
    zoom_range=0.1,
    horizontal_flip=False,
    fill_mode='nearest'
)

    test_datagen = ImageDataGenerator(rescale=1./255)

    train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224),
        batch_size=16,
        class_mode='binary',
        shuffle=True
    )
    test_generator = test_datagen.flow_from_directory(
        test_dir,
        target_size=(224, 224),
        batch_size=16,
        class_mode='binary',
        shuffle=False
    )

    # Build model with MobileNetV2 base
    base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
    base_model.trainable = False

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu', kernel_regularizer='l2'),
        layers.Dropout(0.6),
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    model.summary()

    callbacks = [
        EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True),
        ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=4, verbose=1)
    ]

    history = model.fit(
        train_generator,
        validation_data=test_generator,
        epochs=40,
        callbacks=callbacks
    )

    loss, acc = model.evaluate(test_generator, verbose=0)
    print(f"\n✅ Final Test Accuracy on {dataset_name}: {acc:.2f}")

    # Plot training curves
    plt.figure(figsize=(12,5))
    plt.subplot(1,2,1)
    plt.plot(history.history['accuracy'], label="Train Acc")
    plt.plot(history.history['val_accuracy'], label="Val Acc")
    plt.title(f"{dataset_name} Model Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(history.history['loss'], label="Train Loss")
    plt.plot(history.history['val_loss'], label="Val Loss")
    plt.title(f"{dataset_name} Model Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()

    plt.show()

# Train and evaluate on both datasets
for ds in ['spiral', 'wave']:
    prepare_and_train(ds)
```

```python
# === Cell A: Grad-CAM helper functions ===
import os
import numpy as np
import tensorflow as tf
import cv2
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image
from PIL import Image

def find_last_conv_layer_name(model):
    base = model.layers[0] if hasattr(model, "layers") and len(model.layers) and hasattr(model.layers[0], "layers") else model
    last_conv = None
    if hasattr(base, "layers"):
        for layer in reversed(base.layers):
            if isinstance(layer, tf.keras.layers.Conv2D):
                last_conv = layer.name
                break
    if last_conv is None:
        for layer in reversed(model.layers):
            if isinstance(layer, tf.keras.layers.Conv2D):
                last_conv = layer.name
                break
    if last_conv is None:
        raise ValueError("No Conv2D layer found.")
    return last_conv

def make_gradcam_heatmap(img_array, model, last_conv_layer_name, eps=1e-8):
    base = model.layers[0] if hasattr(model, "layers") and hasattr(model.layers[0], "layers") else model
    try:
        last_conv_layer = base.get_layer(last_conv_layer_name).output
    except Exception:
        last_conv_layer = model.get_layer(last_conv_layer_name).output
    grad_model = tf.keras.models.Model([model.inputs], [last_conv_layer, model.output])
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        loss = predictions[:, 0]
    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_outputs = conv_outputs[0]
    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
    heatmap = tf.nn.relu(heatmap)
    heatmap = heatmap / (tf.reduce_max(heatmap) + eps)
    return heatmap.numpy()

def apply_heatmap_on_image(orig_bgr, heatmap, alpha=0.45):
    heatmap_uint8 = np.uint8(255 * heatmap)
    heatmap_resized = cv2.resize(heatmap_uint8, (orig_bgr.shape[1], orig_bgr.shape[0]))
    heatmap_color = cv2.applyColorMap(heatmap_resized, cv2.COLORMAP_JET)
    overlay = heatmap_color.astype(np.float32) * alpha + orig_bgr.astype(np.float32)
    return np.clip(overlay, 0, 255).astype(np.uint8)

def pil_to_bgr(pil_img):
    arr = np.array(pil_img)
    if arr.ndim == 2:
        arr = np.stack([arr,arr,arr], axis=-1)
    return cv2.cvtColor(arr, cv2.COLOR_RGB2BGR)
```

```python
# ===== Grad-CAM for Top 5 Parkinson Images =====
import os, glob
import numpy as np
import cv2
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing import image
from PIL import UnidentifiedImageError

# ---------- Helper Functions ----------
def find_last_conv_layer_name(model):
    for layer in reversed(model.layers):
        if isinstance(layer, tf.keras.layers.Conv2D):
            return layer.name
        elif hasattr(layer, 'layers'):
            try:
                return find_last_conv_layer_name(layer)
            except ValueError:
                continue
    raise ValueError("No Conv2D layer found in the model.")

def pil_to_bgr(pil_img):
    return cv2.cvtColor(np.array(pil_img), cv2.COLOR_RGB2BGR)

def apply_heatmap_on_image(img, heatmap, alpha=0.4, colormap=cv2.COLORMAP_JET):
    heatmap = cv2.resize(heatmap, (img.shape[1], img.shape[0]))
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, colormap)
    overlay = cv2.addWeighted(img, 1 - alpha, heatmap, alpha, 0)
    return overlay

# ---------- Auto-Detect Parkinson Folder ----------
parkinson_dirs = []
for root, dirs, files in os.walk("/content"):
    if "parkinson" in root.lower() and any(f.lower().endswith(('.png', '.jpg', '.jpeg')) for f in files):
        parkinson_dirs.append(root)

if not parkinson_dirs:
    raise FileNotFoundError("⚠️ No folder containing Parkinson images found!")

parkinson_dir = parkinson_dirs[0]
print(f"✅ Using Parkinson folder: {parkinson_dir}")

img_paths = glob.glob(os.path.join(parkinson_dir, "*"))
if len(img_paths) == 0:
    raise FileNotFoundError("⚠️ No images found in detected folder!")

# ---------- Prepare Model ----------
base_model = model.layers[0]
top_layers = model.layers[1:]
last_conv = find_last_conv_layer_name(base_model)
print(f"✅ Base model: {base_model.name}")
print(f"✅ Last Conv layer: {last_conv}")

last_conv_output = base_model.get_layer(last_conv).output
grad_model = tf.keras.models.Model([base_model.input], [last_conv_output, base_model.output])

# ---------- Get Predictions & Sort by Confidence ----------
predictions = []
for img_path in img_paths:
    try:
        pil_img = image.load_img(img_path, target_size=(224,224))
        img_arr = image.img_to_array(pil_img) / 255.0
        input_batch = np.expand_dims(img_arr, axis=0)

        preds = model.predict(input_batch, verbose=0)
        conf = preds[0][0] if preds.shape[-1] == 1 else preds[0][1]
        predictions.append((img_path, conf))
    except (UnidentifiedImageError, Exception) as e:
        print(f"⚠️ Skipping {img_path} ({e})")

# take top 5
predictions = sorted(predictions, key=lambda x: x[1], reverse=True)[:5]
print(f"Selected {len(predictions)} most confident Parkinson images.")

# ---------- Grad-CAM Visualization ----------
plt.figure(figsize=(20, 8))

for i, (img_path, _) in enumerate(predictions):
    pil_img = image.load_img(img_path, target_size=(224,224))
    img_arr = image.img_to_array(pil_img) / 255.0
    input_batch = np.expand_dims(img_arr, axis=0)

    with tf.GradientTape() as tape:
        conv_outputs, features = grad_model(input_batch, training=False)
        tape.watch(conv_outputs)
        x = features
        for layer in top_layers:
            x = layer(x)
        pred = tf.sigmoid(x)

    grads = tape.gradient(pred, conv_outputs)
    if grads is None:
        continue

    grads = grads[0] / (tf.reduce_max(tf.abs(grads)) + 1e-8)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1))

    conv_outputs = conv_outputs[0]
    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
    heatmap = tf.nn.relu(heatmap)
    heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)
    heatmap = heatmap.numpy()

    orig_bgr = cv2.imread(img_path)
    if orig_bgr is None:
        orig_bgr = pil_to_bgr(pil_img)

    overlay = apply_heatmap_on_image(orig_bgr, heatmap)

    # plot each of the 5 images separately
    plt.subplot(1, 5, i+1)
    plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
    plt.title(f"Image {i+1}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
